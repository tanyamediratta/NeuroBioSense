```python
# ============================================================
# FULL NEUROBIOSENSE PIPELINE
# 1. Video Emotion Model: 3D CNN
# 2. Biosignal Emotion Model: 1D CNN
# 3. Fusion Difference Model: 3D CNN + 1D CNN
# ============================================================

!pip install -q opencv-python scikit-learn

import os
import cv2
import json
import time
import copy
import math
import random
import hashlib
import warnings
import numpy as np
import pandas as pd

from pathlib import Path
from collections import Counter

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from torchvision.models.video import r3d_18, R3D_18_Weights
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

warnings.filterwarnings("ignore")

# ============================================================
# CONFIG
# ============================================================

ROOT = "/content/drive/MyDrive/NeuroBioSense_Dataset/NeuroBioSense"
ADS_DIR = f"{ROOT}/Advertisement Categories"
BIO_CSV = f"{ROOT}/Biosignal Files/Pre-Processed/4-Hertz.csv"

OUT_DIR = "/content/drive/MyDrive/NeuroBioSense_Outputs"
os.makedirs(OUT_DIR, exist_ok=True)

SEED = 42
BIO_FS = 4

N_FRAMES = 16
FRAME_SIZE = 112
BIO_SEQ_LEN = 128

BATCH_VIDEO = 4
BATCH_BIO = 32
BATCH_FUSION = 4

VIDEO_EPOCHS = 12
BIO_EPOCHS = 25
FUSION_EPOCHS = 12

LR_VIDEO = 1e-4
LR_BIO = 1e-3
LR_FUSION = 1e-4

TEST_SIZE = 0.2
NUM_CLASSES = 7

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

EMOTION_MAP = {
    "N": 0,
    "A": 1,
    "D": 2,
    "F": 3,
    "J": 4,
    "SA": 5,
    "SU": 6,
}

ID_TO_EMOTION = {v: k for k, v in EMOTION_MAP.items()}

EMOTION_NAMES = {
    "N": "Neutral",
    "A": "Anger",
    "D": "Disgust",
    "F": "Fear",
    "J": "Joy",
    "SA": "Sadness",
    "SU": "Surprise",
}

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(SEED)

# ============================================================
# LABEL HELPERS
# ============================================================

def normalize_emotion_label(value):
    if value is None or pd.isna(value):
        return None

    text = str(value).strip().upper().replace(" ", "")

    fixes = {
        "NEUTRAL": "N",
        "ANGER": "A",
        "ANGRY": "A",
        "DISGUST": "D",
        "FEAR": "F",
        "JOY": "J",
        "HAPPY": "J",
        "SAD": "SA",
        "SADNESS": "SA",
        "SURPRISE": "SU",
        "SURPRSIE": "SU",
        "SURPRISED": "SU",
        "J1": "J",
    }

    text = fixes.get(text, text)
    return EMOTION_MAP.get(text)

def label_name(i):
    code = ID_TO_EMOTION[int(i)]
    return f"{code}-{EMOTION_NAMES[code]}"

# ============================================================
# COLLECT VIDEO METADATA
# ============================================================

def stable_video_id(path):
    rel = os.path.relpath(path, ADS_DIR)
    return hashlib.md5(rel.encode("utf-8")).hexdigest()

def collect_videos():
    rows = []

    for category in sorted(os.listdir(ADS_DIR)):
        if category.startswith("."):
            continue

        category_dir = os.path.join(ADS_DIR, category)
        if not os.path.isdir(category_dir):
            continue

        for root, dirs, files in os.walk(category_dir):
            dirs[:] = [d for d in dirs if not d.startswith(".")]

            for f in files:
                if f.startswith("."):
                    continue

                if not f.lower().endswith(".mp4"):
                    continue

                path = os.path.join(root, f)
                parent_folder = os.path.basename(os.path.dirname(path))
                video_label = normalize_emotion_label(parent_folder)

                if video_label is None:
                    continue

                cap = cv2.VideoCapture(path)
                fps = cap.get(cv2.CAP_PROP_FPS)
                n_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
                cap.release()

                if fps is None or fps <= 1 or n_frames <= 0:
                    continue

                rows.append({
                    "video_id": stable_video_id(path),
                    "path": path,
                    "category": category,
                    "ad_folder": parent_folder,
                    "video_label": video_label,
                    "fps": float(fps),
                    "n_frames": int(n_frames),
                    "duration_sec": float(n_frames / fps),
                })

    df = pd.DataFrame(rows)
    print("Videos collected:", len(df))
    print("Video label counts:", Counter(df["video_label"]))
    return df


meta = collect_videos()
display(meta.head())

# ============================================================
# LOAD BIOSIGNAL DATA
# ============================================================

bio_df = pd.read_csv(BIO_CSV)
print("Bio CSV shape:", bio_df.shape)
print("Bio columns:", bio_df.columns.tolist()[:30])

if "EMOTION" not in bio_df.columns:
    raise ValueError("BIO_CSV must contain an EMOTION column.")

bio_df["emotion_id"] = bio_df["EMOTION"].apply(normalize_emotion_label)

if bio_df["emotion_id"].isna().any():
    bad = bio_df[bio_df["emotion_id"].isna()]["EMOTION"].unique()
    raise ValueError(f"Unknown biosignal emotion labels: {bad}")

bio_df["emotion_id"] = bio_df["emotion_id"].astype(int)

# Use real numeric biosignal columns as input.
# Exclude label/meta columns.
exclude_cols = {
    "EMOTION",
    "emotion_id",
    "Emotion",
    "emotion",
    "Label",
    "label",
    "Time",
    "time",
    "Timestamp",
    "timestamp",
}

numeric_cols = [
    c for c in bio_df.select_dtypes(include=[np.number]).columns
    if c not in exclude_cols
]

if len(numeric_cols) == 0:
    raise ValueError("No numeric biosignal feature columns found.")

print("Using biosignal feature columns:", numeric_cols)
bio_values = bio_df[numeric_cols].values.astype(np.float32)
bio_emotions = bio_df["emotion_id"].values.astype(np.int64)

# ============================================================
# DETERMINISTIC VIDEO → BIOSIGNAL LINKING
# ============================================================

def hash_start(video_id, pool_size, slice_len, seed="nbs-2026"):
    h = hashlib.sha256(f"{seed}|{video_id}".encode("utf-8")).digest()
    big = int.from_bytes(h[:8], "big")
    return big % max(1, pool_size - slice_len - 1)

def build_linking(meta_df):
    rows = []
    pool_size = len(bio_df)

    for _, r in meta_df.iterrows():
        slice_len = max(4, int(round(r["duration_sec"] * BIO_FS)))
        start = hash_start(r["video_id"], pool_size, slice_len)
        end = start + slice_len

        bio_slice_labels = bio_emotions[start:end]
        bio_label = int(np.bincount(bio_slice_labels, minlength=NUM_CLASSES).argmax())

        rows.append({
            "video_id": r["video_id"],
            "path": r["path"],
            "category": r["category"],
            "ad_folder": r["ad_folder"],
            "video_label": int(r["video_label"]),
            "bio_label": bio_label,
            "bio_start": int(start),
            "bio_end": int(end),
            "duration_sec": float(r["duration_sec"]),
        })

    link_df = pd.DataFrame(rows)
    print("Linked samples:", len(link_df))
    print("Video labels:", Counter(link_df["video_label"]))
    print("Bio labels:", Counter(link_df["bio_label"]))
    print("Difference labels:", Counter((link_df["video_label"] != link_df["bio_label"]).astype(int)))

    link_df.to_csv(os.path.join(OUT_DIR, "linking_table_three_model.csv"), index=False)
    return link_df

link_df = build_linking(meta)
display(link_df.head())

# ============================================================
# TRAIN/TEST SPLIT BY VIDEO ID
# ============================================================

gss = GroupShuffleSplit(n_splits=1, test_size=TEST_SIZE, random_state=SEED)
train_idx, test_idx = next(gss.split(link_df, groups=link_df["video_id"]))

train_df = link_df.iloc[train_idx].reset_index(drop=True)
test_df = link_df.iloc[test_idx].reset_index(drop=True)

overlap = set(train_df["video_id"]) & set(test_df["video_id"])
print("Train videos:", len(train_df))
print("Test videos:", len(test_df))
print("Overlap:", len(overlap))

# ============================================================
# NORMALIZE BIOSIGNAL USING TRAIN DATA ONLY
# ============================================================

train_bio_parts = []
for _, r in train_df.iterrows():
    train_bio_parts.append(bio_values[int(r["bio_start"]):int(r["bio_end"])])

train_bio_concat = np.concatenate(train_bio_parts, axis=0)
BIO_MEAN = train_bio_concat.mean(axis=0, keepdims=True)
BIO_STD = train_bio_concat.std(axis=0, keepdims=True) + 1e-6

np.save(os.path.join(OUT_DIR, "bio_mean.npy"), BIO_MEAN)
np.save(os.path.join(OUT_DIR, "bio_std.npy"), BIO_STD)

# ============================================================
# VIDEO LOADING
# ============================================================

KIN_MEAN = torch.tensor([0.43216, 0.394666, 0.37645]).view(3, 1, 1, 1)
KIN_STD = torch.tensor([0.22803, 0.22145, 0.216989]).view(3, 1, 1, 1)

def read_clip(path, n_frames=N_FRAMES, size=FRAME_SIZE, mode="random", clip_index=0, total_clips=3):
    cap = cv2.VideoCapture(path)
    total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    if total <= 0:
        cap.release()
        return torch.zeros(3, n_frames, size, size)

    if total < n_frames:
        frame_indices = np.linspace(0, max(0, total - 1), n_frames).astype(int)
    else:
        max_start = total - n_frames

        if mode == "random":
            start = random.randint(0, max_start)
        else:
            if total_clips <= 1:
                start = max_start // 2
            else:
                starts = np.linspace(0, max_start, total_clips).astype(int)
                start = int(starts[min(clip_index, total_clips - 1)])

        frame_indices = np.arange(start, start + n_frames)

    frames = []
    current = 0
    wanted = set(frame_indices.tolist())

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        if current in wanted:
            frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            frame = cv2.resize(frame, (size, size), interpolation=cv2.INTER_AREA)
            frames.append(frame)

        current += 1
        if len(frames) == n_frames:
            break

    cap.release()

    if len(frames) == 0:
        return torch.zeros(3, n_frames, size, size)

    while len(frames) < n_frames:
        frames.append(frames[-1])

    arr = np.stack(frames, axis=0)
    x = torch.from_numpy(arr).float().permute(3, 0, 1, 2) / 255.0
    x = (x - KIN_MEAN) / KIN_STD
    return x

# ============================================================
# BIOSIGNAL LOADING
# ============================================================

def resample_bio_sequence(seq, seq_len=BIO_SEQ_LEN):
    # seq shape: T x C
    if len(seq) == 0:
        return np.zeros((seq_len, bio_values.shape[1]), dtype=np.float32)

    if len(seq) == 1:
        return np.repeat(seq, seq_len, axis=0).astype(np.float32)

    old_x = np.linspace(0, 1, len(seq))
    new_x = np.linspace(0, 1, seq_len)

    out = []
    for c in range(seq.shape[1]):
        out.append(np.interp(new_x, old_x, seq[:, c]))

    return np.stack(out, axis=1).astype(np.float32)

def get_bio_tensor(start, end):
    seq = bio_values[int(start):int(end)]
    seq = (seq - BIO_MEAN) / BIO_STD
    seq = resample_bio_sequence(seq, BIO_SEQ_LEN)
    # Return C x T for Conv1D
    return torch.tensor(seq.T, dtype=torch.float32)

# ============================================================
# DATASETS
# ============================================================

class VideoDataset(Dataset):
    def __init__(self, df, train=True):
        self.df = df.reset_index(drop=True)
        self.train = train

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        r = self.df.iloc[idx]
        clip = read_clip(r["path"], mode="random" if self.train else "eval")
        y = int(r["video_label"])
        return clip, torch.tensor(y, dtype=torch.long)

class BioDataset(Dataset):
    def __init__(self, df):
        self.df = df.reset_index(drop=True)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        r = self.df.iloc[idx]
        x = get_bio_tensor(r["bio_start"], r["bio_end"])
        y = int(r["bio_label"])
        return x, torch.tensor(y, dtype=torch.long)

class FusionDataset(Dataset):
    def __init__(self, df, train=True):
        self.df = df.reset_index(drop=True)
        self.train = train

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        r = self.df.iloc[idx]

        clip = read_clip(r["path"], mode="random" if self.train else "eval")
        bio = get_bio_tensor(r["bio_start"], r["bio_end"])

        # 0 = Match, 1 = Mismatch
        diff = int(int(r["video_label"]) != int(r["bio_label"]))

        return clip, bio, torch.tensor(diff, dtype=torch.long)

# ============================================================
# MODELS
# ============================================================

class Video3DCNN(nn.Module):
    def __init__(self, num_classes=7):
        super().__init__()
        weights = R3D_18_Weights.DEFAULT
        self.backbone = r3d_18(weights=weights)

        in_features = self.backbone.fc.in_features
        self.backbone.fc = nn.Identity()

        # Freeze early layers for stability/speed
        for name, p in self.backbone.named_parameters():
            if name.startswith("stem") or name.startswith("layer1"):
                p.requires_grad = False

        self.classifier = nn.Sequential(
            nn.Dropout(0.35),
            nn.Linear(in_features, num_classes)
        )

    def features(self, x):
        return self.backbone(x)

    def forward(self, x):
        feat = self.features(x)
        return self.classifier(feat)

class Bio1DCNN(nn.Module):
    def __init__(self, in_channels, num_classes=7):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Conv1d(in_channels, 64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(64, 128, kernel_size=5, padding=2),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm1d(128),
            nn.ReLU(),

            nn.AdaptiveAvgPool1d(1)
        )

        self.classifier = nn.Sequential(
            nn.Dropout(0.30),
            nn.Linear(128, num_classes)
        )

    def features(self, x):
        x = self.encoder(x).squeeze(-1)
        return x

    def forward(self, x):
        feat = self.features(x)
        return self.classifier(feat)

class FusionDifferenceModel(nn.Module):
    def __init__(self, video_model, bio_model):
        super().__init__()

        self.video_model = video_model
        self.bio_model = bio_model

        # Freeze trained encoders initially
        for p in self.video_model.parameters():
            p.requires_grad = False
        for p in self.bio_model.parameters():
            p.requires_grad = False

        self.classifier = nn.Sequential(
            nn.Linear(512 + 128 + 7 + 7, 256),
            nn.ReLU(),
            nn.Dropout(0.35),
            nn.Linear(256, 64),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(64, 2)
        )

    def forward(self, video, bio):
        video_feat = self.video_model.features(video)
        bio_feat = self.bio_model.features(bio)

        video_logits = self.video_model.classifier(video_feat)
        bio_logits = self.bio_model.classifier(bio_feat)

        video_probs = torch.softmax(video_logits, dim=1)
        bio_probs = torch.softmax(bio_logits, dim=1)

        fused = torch.cat([video_feat, bio_feat, video_probs, bio_probs], dim=1)
        return self.classifier(fused)

# ============================================================
# TRAINING HELPERS
# ============================================================

def class_weights(labels, num_classes):
    counts = np.bincount(np.array(labels), minlength=num_classes).astype(np.float32)
    counts[counts == 0] = 1.0
    weights = counts.sum() / (num_classes * counts)
    return torch.tensor(weights, dtype=torch.float32).to(DEVICE)

def train_simple_model(model, train_loader, val_loader, train_labels, num_classes, epochs, lr, name):
    model = model.to(DEVICE)

    optimizer = torch.optim.AdamW(
        filter(lambda p: p.requires_grad, model.parameters()),
        lr=lr,
        weight_decay=1e-4
    )

    criterion = nn.CrossEntropyLoss(weight=class_weights(train_labels, num_classes))
    scaler = torch.cuda.amp.GradScaler(enabled=torch.cuda.is_available())

    best_acc = 0
    best_state = None

    for epoch in range(1, epochs + 1):
        model.train()
        losses = []

        for batch in train_loader:
            optimizer.zero_grad()

            if len(batch) == 2:
                x, y = batch
                x = x.to(DEVICE)
                y = y.to(DEVICE)

                with torch.cuda.amp.autocast(enabled=torch.cuda.is_available()):
                    logits = model(x)
                    loss = criterion(logits, y)

            else:
                video, bio, y = batch
                video = video.to(DEVICE)
                bio = bio.to(DEVICE)
                y = y.to(DEVICE)

                with torch.cuda.amp.autocast(enabled=torch.cuda.is_available()):
                    logits = model(video, bio)
                    loss = criterion(logits, y)

            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()

            losses.append(loss.item())

        val_acc = evaluate_accuracy(model, val_loader)
        print(f"{name} Epoch {epoch:02d}/{epochs} | Loss {np.mean(losses):.4f} | Val Acc {val_acc:.4f}")

        if val_acc > best_acc:
            best_acc = val_acc
            best_state = copy.deepcopy(model.state_dict())

    if best_state is not None:
        model.load_state_dict(best_state)

    print(f"Best {name} val acc: {best_acc:.4f}")
    return model

def evaluate_accuracy(model, loader):
    model.eval()
    y_true, y_pred = [], []

    with torch.no_grad():
        for batch in loader:
            if len(batch) == 2:
                x, y = batch
                x = x.to(DEVICE)
                logits = model(x)
            else:
                video, bio, y = batch
                video = video.to(DEVICE)
                bio = bio.to(DEVICE)
                logits = model(video, bio)

            pred = logits.argmax(dim=1).cpu().numpy()
            y_true.extend(y.numpy().tolist())
            y_pred.extend(pred.tolist())

    return accuracy_score(y_true, y_pred)

def full_report(model, loader, title, num_classes=7, target_names=None):
    model.eval()
    y_true, y_pred = [], []

    with torch.no_grad():
        for batch in loader:
            if len(batch) == 2:
                x, y = batch
                x = x.to(DEVICE)
                logits = model(x)
            else:
                video, bio, y = batch
                video = video.to(DEVICE)
                bio = bio.to(DEVICE)
                logits = model(video, bio)

            pred = logits.argmax(dim=1).cpu().numpy()
            y_true.extend(y.numpy().tolist())
            y_pred.extend(pred.tolist())

    print("\n" + "=" * 60)
    print(title)
    print("=" * 60)
    print("Accuracy:", accuracy_score(y_true, y_pred))

    if target_names is None:
        target_names = [label_name(i) for i in range(num_classes)]

    print(classification_report(
        y_true,
        y_pred,
        labels=list(range(num_classes)),
        target_names=target_names,
        zero_division=0
    ))

    print("Confusion matrix:")
    print(confusion_matrix(y_true, y_pred, labels=list(range(num_classes))))

    return y_true, y_pred

# ============================================================
# DATALOADERS
# ============================================================

video_train_loader = DataLoader(
    VideoDataset(train_df, train=True),
    batch_size=BATCH_VIDEO,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

video_test_loader = DataLoader(
    VideoDataset(test_df, train=False),
    batch_size=BATCH_VIDEO,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

bio_train_loader = DataLoader(
    BioDataset(train_df),
    batch_size=BATCH_BIO,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

bio_test_loader = DataLoader(
    BioDataset(test_df),
    batch_size=BATCH_BIO,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

fusion_train_loader = DataLoader(
    FusionDataset(train_df, train=True),
    batch_size=BATCH_FUSION,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

fusion_test_loader = DataLoader(
    FusionDataset(test_df, train=False),
    batch_size=BATCH_FUSION,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

# ============================================================
# PHASE 1: TRAIN VIDEO 3D CNN MODEL
# ============================================================

print("\n\nPHASE 1: Training Video 3D CNN Emotion Model")

video_model = Video3DCNN(num_classes=NUM_CLASSES)

video_model = train_simple_model(
    model=video_model,
    train_loader=video_train_loader,
    val_loader=video_test_loader,
    train_labels=train_df["video_label"].values,
    num_classes=NUM_CLASSES,
    epochs=VIDEO_EPOCHS,
    lr=LR_VIDEO,
    name="Video 3D CNN"
)

full_report(
    video_model,
    video_test_loader,
    "FINAL VIDEO EMOTION MODEL",
    num_classes=NUM_CLASSES,
    target_names=[label_name(i) for i in range(NUM_CLASSES)]
)

torch.save({
    "model_state_dict": video_model.state_dict(),
    "emotion_map": EMOTION_MAP,
    "frame_size": FRAME_SIZE,
    "n_frames": N_FRAMES,
    "model_type": "r3d_18_video_emotion"
}, os.path.join(OUT_DIR, "video_3dcnn_emotion.pt"))

print("Saved:", os.path.join(OUT_DIR, "video_3dcnn_emotion.pt"))

# ============================================================
# PHASE 2: TRAIN BIOSIGNAL 1D CNN MODEL
# ============================================================

print("\n\nPHASE 2: Training Biosignal 1D CNN Emotion Model")

bio_model = Bio1DCNN(
    in_channels=len(numeric_cols),
    num_classes=NUM_CLASSES
)

bio_model = train_simple_model(
    model=bio_model,
    train_loader=bio_train_loader,
    val_loader=bio_test_loader,
    train_labels=train_df["bio_label"].values,
    num_classes=NUM_CLASSES,
    epochs=BIO_EPOCHS,
    lr=LR_BIO,
    name="Biosignal 1D CNN"
)

full_report(
    bio_model,
    bio_test_loader,
    "FINAL BIOSIGNAL EMOTION MODEL",
    num_classes=NUM_CLASSES,
    target_names=[label_name(i) for i in range(NUM_CLASSES)]
)

torch.save({
    "model_state_dict": bio_model.state_dict(),
    "emotion_map": EMOTION_MAP,
    "bio_columns": numeric_cols,
    "bio_seq_len": BIO_SEQ_LEN,
    "bio_mean_path": "bio_mean.npy",
    "bio_std_path": "bio_std.npy",
    "model_type": "cnn1d_biosignal_emotion"
}, os.path.join(OUT_DIR, "bio_1dcnn_emotion.pt"))

print("Saved:", os.path.join(OUT_DIR, "bio_1dcnn_emotion.pt"))

# ============================================================
# PHASE 3: TRAIN FUSION DIFFERENCE MODEL
# ============================================================

print("\n\nPHASE 3: Training Fusion Difference Model")

fusion_model = FusionDifferenceModel(video_model, bio_model)

fusion_train_labels = (train_df["video_label"].values != train_df["bio_label"].values).astype(int)

fusion_model = train_simple_model(
    model=fusion_model,
    train_loader=fusion_train_loader,
    val_loader=fusion_test_loader,
    train_labels=fusion_train_labels,
    num_classes=2,
    epochs=FUSION_EPOCHS,
    lr=LR_FUSION,
    name="Fusion Difference"
)

full_report(
    fusion_model,
    fusion_test_loader,
    "FINAL FUSION DIFFERENCE MODEL",
    num_classes=2,
    target_names=["Match", "Mismatch"]
)

torch.save({
    "model_state_dict": fusion_model.state_dict(),
    "emotion_map": EMOTION_MAP,
    "bio_columns": numeric_cols,
    "bio_seq_len": BIO_SEQ_LEN,
    "frame_size": FRAME_SIZE,
    "n_frames": N_FRAMES,
    "model_type": "fusion_3dcnn_1dcnn_match_mismatch"
}, os.path.join(OUT_DIR, "fusion_3dcnn_1dcnn_difference.pt"))

print("Saved:", os.path.join(OUT_DIR, "fusion_3dcnn_1dcnn_difference.pt"))

# ============================================================
# SAVE FINAL METADATA
# ============================================================

final_info = {
    "root": ROOT,
    "output_dir": OUT_DIR,
    "models": {
        "video_model": "video_3dcnn_emotion.pt",
        "bio_model": "bio_1dcnn_emotion.pt",
        "fusion_model": "fusion_3dcnn_1dcnn_difference.pt",
    },
    "emotion_map": EMOTION_MAP,
    "bio_columns": numeric_cols,
    "n_frames": N_FRAMES,
    "frame_size": FRAME_SIZE,
    "bio_seq_len": BIO_SEQ_LEN,
    "train_videos": len(train_df),
    "test_videos": len(test_df),
}

with open(os.path.join(OUT_DIR, "final_project_metadata.json"), "w") as f:
    json.dump(final_info, f, indent=2)

print("\n\nDONE.")
print("All outputs saved to:", OUT_DIR)
print("Saved files:")
print(os.listdir(OUT_DIR))
```

```python
# ============================================================
# FAST FIX: BALANCED FUSION MODEL
# Uses already trained video_3dcnn_emotion.pt and bio_1dcnn_emotion.pt
# Trains only a lightweight fusion classifier with balanced sampling
# ============================================================

import os
import json
import copy
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F

from collections import Counter
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, balanced_accuracy_score, f1_score
from torchvision.models.video import r3d_18, R3D_18_Weights

# ============================================================
# CONFIG
# ============================================================

OUT_DIR = "/content/drive/MyDrive/NeuroBioSense_Outputs"

LINK_CSV = os.path.join(OUT_DIR, "linking_table_three_model.csv")
VIDEO_MODEL_PATH = os.path.join(OUT_DIR, "video_3dcnn_emotion.pt")
BIO_MODEL_PATH = os.path.join(OUT_DIR, "bio_1dcnn_emotion.pt")
BIO_MEAN_PATH = os.path.join(OUT_DIR, "bio_mean.npy")
BIO_STD_PATH = os.path.join(OUT_DIR, "bio_std.npy")

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

SEED = 42
N_FRAMES = 16
FRAME_SIZE = 112
BIO_SEQ_LEN = 128
BATCH = 8
FUSION_EPOCHS = 25
LR = 1e-3

NUM_CLASSES = 7

EMOTION_MAP = {
    "N": 0,
    "A": 1,
    "D": 2,
    "F": 3,
    "J": 4,
    "SA": 5,
    "SU": 6,
}

ID_TO_EMOTION = {v: k for k, v in EMOTION_MAP.items()}

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

# ============================================================
# LOAD TABLE
# ============================================================

df = pd.read_csv(LINK_CSV)
df["diff_label"] = (df["video_label"].astype(int) != df["bio_label"].astype(int)).astype(int)

print("Total samples:", len(df))
print("Diff counts:", Counter(df["diff_label"]))

# Same grouped split as before
from sklearn.model_selection import GroupShuffleSplit

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=SEED)
train_idx, test_idx = next(gss.split(df, groups=df["video_id"]))

train_df = df.iloc[train_idx].reset_index(drop=True)
test_df = df.iloc[test_idx].reset_index(drop=True)

print("Train diff counts:", Counter(train_df["diff_label"]))
print("Test diff counts:", Counter(test_df["diff_label"]))

# ============================================================
# LOAD NORMALIZATION
# ============================================================

BIO_MEAN = np.load(BIO_MEAN_PATH)
BIO_STD = np.load(BIO_STD_PATH)

ROOT = "/content/drive/MyDrive/NeuroBioSense_Dataset/NeuroBioSense"
BIO_CSV = f"{ROOT}/Biosignal Files/Pre-Processed/4-Hertz.csv"

bio_df = pd.read_csv(BIO_CSV)
numeric_cols = ["BVP", "EDA", "TEMP", "X", "Y", "Z"]
bio_values = bio_df[numeric_cols].values.astype(np.float32)

# ============================================================
# VIDEO + BIO LOADERS
# ============================================================

KIN_MEAN = torch.tensor([0.43216, 0.394666, 0.37645]).view(3, 1, 1, 1)
KIN_STD = torch.tensor([0.22803, 0.22145, 0.216989]).view(3, 1, 1, 1)

def read_clip(path, n_frames=N_FRAMES, size=FRAME_SIZE):
    cap = cv2.VideoCapture(path)
    total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    if total <= 0:
        cap.release()
        return torch.zeros(3, n_frames, size, size)

    if total < n_frames:
        frame_indices = np.linspace(0, max(0, total - 1), n_frames).astype(int)
    else:
        start = max(0, (total - n_frames) // 2)
        frame_indices = np.arange(start, start + n_frames)

    frames = []
    current = 0
    wanted = set(frame_indices.tolist())

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        if current in wanted:
            frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            frame = cv2.resize(frame, (size, size), interpolation=cv2.INTER_AREA)
            frames.append(frame)

        current += 1
        if len(frames) == n_frames:
            break

    cap.release()

    if len(frames) == 0:
        return torch.zeros(3, n_frames, size, size)

    while len(frames) < n_frames:
        frames.append(frames[-1])

    arr = np.stack(frames, axis=0)
    x = torch.from_numpy(arr).float().permute(3, 0, 1, 2) / 255.0
    x = (x - KIN_MEAN) / KIN_STD
    return x

def resample_bio_sequence(seq, seq_len=BIO_SEQ_LEN):
    if len(seq) == 0:
        return np.zeros((seq_len, bio_values.shape[1]), dtype=np.float32)

    if len(seq) == 1:
        return np.repeat(seq, seq_len, axis=0).astype(np.float32)

    old_x = np.linspace(0, 1, len(seq))
    new_x = np.linspace(0, 1, seq_len)

    out = []
    for c in range(seq.shape[1]):
        out.append(np.interp(new_x, old_x, seq[:, c]))

    return np.stack(out, axis=1).astype(np.float32)

def get_bio_tensor(start, end):
    seq = bio_values[int(start):int(end)]
    seq = (seq - BIO_MEAN) / BIO_STD
    seq = resample_bio_sequence(seq, BIO_SEQ_LEN)
    return torch.tensor(seq.T, dtype=torch.float32)

class ExtractDataset(Dataset):
    def __init__(self, df):
        self.df = df.reset_index(drop=True)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        r = self.df.iloc[idx]
        video = read_clip(r["path"])
        bio = get_bio_tensor(r["bio_start"], r["bio_end"])
        y = int(r["diff_label"])
        return video, bio, torch.tensor(y, dtype=torch.long)

# ============================================================
# MODEL DEFINITIONS
# Must match previous code
# ============================================================

class Video3DCNN(nn.Module):
    def __init__(self, num_classes=7):
        super().__init__()
        self.backbone = r3d_18(weights=R3D_18_Weights.DEFAULT)
        in_features = self.backbone.fc.in_features
        self.backbone.fc = nn.Identity()

        for name, p in self.backbone.named_parameters():
            if name.startswith("stem") or name.startswith("layer1"):
                p.requires_grad = False

        self.classifier = nn.Sequential(
            nn.Dropout(0.35),
            nn.Linear(in_features, num_classes)
        )

    def features(self, x):
        return self.backbone(x)

    def forward(self, x):
        feat = self.features(x)
        return self.classifier(feat)

class Bio1DCNN(nn.Module):
    def __init__(self, in_channels, num_classes=7):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Conv1d(in_channels, 64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(64, 128, kernel_size=5, padding=2),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm1d(128),
            nn.ReLU(),

            nn.AdaptiveAvgPool1d(1)
        )

        self.classifier = nn.Sequential(
            nn.Dropout(0.30),
            nn.Linear(128, num_classes)
        )

    def features(self, x):
        return self.encoder(x).squeeze(-1)

    def forward(self, x):
        return self.classifier(self.features(x))

# ============================================================
# LOAD TRAINED MODELS
# ============================================================

video_model = Video3DCNN(NUM_CLASSES).to(DEVICE)
video_ckpt = torch.load(VIDEO_MODEL_PATH, map_location=DEVICE)
video_model.load_state_dict(video_ckpt["model_state_dict"])
video_model.eval()

bio_model = Bio1DCNN(in_channels=6, num_classes=NUM_CLASSES).to(DEVICE)
bio_ckpt = torch.load(BIO_MODEL_PATH, map_location=DEVICE)
bio_model.load_state_dict(bio_ckpt["model_state_dict"])
bio_model.eval()

for p in video_model.parameters():
    p.requires_grad = False
for p in bio_model.parameters():
    p.requires_grad = False

# ============================================================
# EXTRACT FEATURES FAST ONCE
# ============================================================

def extract_fusion_features(df, name):
    loader = DataLoader(
        ExtractDataset(df),
        batch_size=BATCH,
        shuffle=False,
        num_workers=2,
        pin_memory=True
    )

    X = []
    y = []

    with torch.no_grad():
        for i, (video, bio, label) in enumerate(loader):
            video = video.to(DEVICE)
            bio = bio.to(DEVICE)

            video_feat = video_model.features(video)
            bio_feat = bio_model.features(bio)

            video_logits = video_model.classifier(video_feat)
            bio_logits = bio_model.classifier(bio_feat)

            video_probs = torch.softmax(video_logits, dim=1)
            bio_probs = torch.softmax(bio_logits, dim=1)

            fused = torch.cat([
                video_feat,
                bio_feat,
                video_probs,
                bio_probs,
                torch.abs(video_probs - bio_probs),
                video_probs * bio_probs
            ], dim=1)

            X.append(fused.cpu().numpy())
            y.append(label.numpy())

            if (i + 1) % 20 == 0:
                print(f"{name}: processed batch {i+1}/{len(loader)}")

    X = np.concatenate(X, axis=0).astype(np.float32)
    y = np.concatenate(y, axis=0).astype(np.int64)

    return X, y

X_train, y_train = extract_fusion_features(train_df, "train")
X_test, y_test = extract_fusion_features(test_df, "test")

print("Fusion feature shape:", X_train.shape)
print("Train labels:", Counter(y_train))
print("Test labels:", Counter(y_test))

# ============================================================
# BALANCED LIGHTWEIGHT FUSION MODEL
# ============================================================

class FusionFeatureDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.long)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

class BalancedFusionMLP(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(0.35),

            nn.Linear(256, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(0.25),

            nn.Linear(64, 2)
        )

    def forward(self, x):
        return self.net(x)

counts = np.bincount(y_train, minlength=2)
sample_weights = np.array([1.0 / counts[y] for y in y_train], dtype=np.float32)

sampler = WeightedRandomSampler(
    weights=sample_weights,
    num_samples=len(sample_weights),
    replacement=True
)

train_loader = DataLoader(
    FusionFeatureDataset(X_train, y_train),
    batch_size=64,
    sampler=sampler
)

test_loader = DataLoader(
    FusionFeatureDataset(X_test, y_test),
    batch_size=128,
    shuffle=False
)

fusion = BalancedFusionMLP(X_train.shape[1]).to(DEVICE)

class_weights = torch.tensor(
    [len(y_train) / (2 * counts[0]), len(y_train) / (2 * counts[1])],
    dtype=torch.float32
).to(DEVICE)

criterion = nn.CrossEntropyLoss(weight=class_weights)
optimizer = torch.optim.AdamW(fusion.parameters(), lr=LR, weight_decay=1e-4)

best_macro_f1 = 0
best_state = None

for epoch in range(1, FUSION_EPOCHS + 1):
    fusion.train()
    losses = []

    for Xb, yb in train_loader:
        Xb = Xb.to(DEVICE)
        yb = yb.to(DEVICE)

        optimizer.zero_grad()
        logits = fusion(Xb)
        loss = criterion(logits, yb)
        loss.backward()
        optimizer.step()

        losses.append(loss.item())

    fusion.eval()
    preds = []
    true = []

    with torch.no_grad():
        for Xb, yb in test_loader:
            logits = fusion(Xb.to(DEVICE))
            pred = logits.argmax(dim=1).cpu().numpy()
            preds.extend(pred.tolist())
            true.extend(yb.numpy().tolist())

    macro_f1 = f1_score(true, preds, average="macro")
    bal_acc = balanced_accuracy_score(true, preds)
    acc = accuracy_score(true, preds)

    print(
        f"Epoch {epoch:02d}/{FUSION_EPOCHS} | "
        f"Loss {np.mean(losses):.4f} | "
        f"Acc {acc:.4f} | BalAcc {bal_acc:.4f} | MacroF1 {macro_f1:.4f}"
    )

    if macro_f1 > best_macro_f1:
        best_macro_f1 = macro_f1
        best_state = copy.deepcopy(fusion.state_dict())

fusion.load_state_dict(best_state)

# ============================================================
# FINAL REPORT
# ============================================================

fusion.eval()
preds = []
true = []

with torch.no_grad():
    for Xb, yb in test_loader:
        logits = fusion(Xb.to(DEVICE))
        pred = logits.argmax(dim=1).cpu().numpy()
        preds.extend(pred.tolist())
        true.extend(yb.numpy().tolist())

print("\nFINAL BALANCED FUSION REPORT")
print("Accuracy:", accuracy_score(true, preds))
print("Balanced Accuracy:", balanced_accuracy_score(true, preds))
print("Macro F1:", f1_score(true, preds, average="macro"))

print(classification_report(
    true,
    preds,
    labels=[0, 1],
    target_names=["Match", "Mismatch"],
    zero_division=0
))

print("Confusion matrix:")
print(confusion_matrix(true, preds, labels=[0, 1]))

torch.save({
    "model_state_dict": fusion.state_dict(),
    "input_dim": X_train.shape[1],
    "model_type": "balanced_fusion_feature_mlp",
    "feature_description": "video_feat + bio_feat + video_probs + bio_probs + absdiff + product"
}, os.path.join(OUT_DIR, "balanced_fusion_fast.pt"))

np.save(os.path.join(OUT_DIR, "fusion_X_train.npy"), X_train)
np.save(os.path.join(OUT_DIR, "fusion_y_train.npy"), y_train)
np.save(os.path.join(OUT_DIR, "fusion_X_test.npy"), X_test)
np.save(os.path.join(OUT_DIR, "fusion_y_test.npy"), y_test)

print("\nSaved:")
print(os.path.join(OUT_DIR, "balanced_fusion_fast.pt"))
```

```python
# ============================================================
# FAST BIOSIGNAL AUGMENTATION + 1D CNN RETRAINING
# Retrains only biosignal model using BVP, EDA, TEMP, X, Y, Z
# Saves: bio_1dcnn_emotion_augmented.pt
# ============================================================

!pip install -q scikit-learn

import os
import json
import copy
import random
import hashlib
import warnings
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F

from collections import Counter
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler

warnings.filterwarnings("ignore")

# ============================================================
# CONFIG
# ============================================================

ROOT = "/content/drive/MyDrive/NeuroBioSense_Dataset/NeuroBioSense"
BIO_CSV = f"{ROOT}/Biosignal Files/Pre-Processed/4-Hertz.csv"
LINK_CSV = "/content/drive/MyDrive/NeuroBioSense_Outputs/linking_table_three_model.csv"
OUT_DIR = "/content/drive/MyDrive/NeuroBioSense_Outputs"

SEED = 42
BIO_SEQ_LEN = 128
BATCH_SIZE = 64
EPOCHS = 40
LR = 1e-3
NUM_CLASSES = 7

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

EMOTION_MAP = {
    "N": 0,
    "A": 1,
    "D": 2,
    "F": 3,
    "J": 4,
    "SA": 5,
    "SU": 6,
}

ID_TO_EMOTION = {v: k for k, v in EMOTION_MAP.items()}

EMOTION_NAMES = {
    "N": "Neutral",
    "A": "Anger",
    "D": "Disgust",
    "F": "Fear",
    "J": "Joy",
    "SA": "Sadness",
    "SU": "Surprise",
}

BIO_COLUMNS = ["BVP", "EDA", "TEMP", "X", "Y", "Z"]

# ============================================================
# LOAD DATA
# ============================================================

bio_df = pd.read_csv(BIO_CSV)
link_df = pd.read_csv(LINK_CSV)

print("Bio CSV:", bio_df.shape)
print("Link table:", link_df.shape)

missing_cols = [c for c in BIO_COLUMNS if c not in bio_df.columns]
if missing_cols:
    raise ValueError(f"Missing biosignal columns: {missing_cols}")

bio_values = bio_df[BIO_COLUMNS].values.astype(np.float32)

# labels from linked table
labels = link_df["bio_label"].astype(int).values
groups = link_df["video_id"].values

print("Bio label counts:", Counter(labels))

# ============================================================
# TRAIN/TEST SPLIT BY VIDEO ID
# ============================================================

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=SEED)
train_idx, test_idx = next(gss.split(link_df, labels, groups=groups))

train_df = link_df.iloc[train_idx].reset_index(drop=True)
test_df = link_df.iloc[test_idx].reset_index(drop=True)

print("Train:", len(train_df), "Test:", len(test_df))
print("Train labels:", Counter(train_df["bio_label"]))
print("Test labels:", Counter(test_df["bio_label"]))

# ============================================================
# NORMALIZE USING TRAIN ONLY
# ============================================================

train_parts = []
for _, r in train_df.iterrows():
    start = int(r["bio_start"])
    end = int(r["bio_end"])
    train_parts.append(bio_values[start:end])

train_concat = np.concatenate(train_parts, axis=0)
BIO_MEAN = train_concat.mean(axis=0, keepdims=True)
BIO_STD = train_concat.std(axis=0, keepdims=True) + 1e-6

np.save(os.path.join(OUT_DIR, "bio_mean_augmented.npy"), BIO_MEAN)
np.save(os.path.join(OUT_DIR, "bio_std_augmented.npy"), BIO_STD)

# ============================================================
# BIOSIGNAL PREPROCESSING
# ============================================================

def resample_bio_sequence(seq, seq_len=BIO_SEQ_LEN):
    if len(seq) == 0:
        return np.zeros((seq_len, len(BIO_COLUMNS)), dtype=np.float32)

    if len(seq) == 1:
        return np.repeat(seq, seq_len, axis=0).astype(np.float32)

    old_x = np.linspace(0, 1, len(seq))
    new_x = np.linspace(0, 1, seq_len)

    out = []
    for c in range(seq.shape[1]):
        out.append(np.interp(new_x, old_x, seq[:, c]))

    return np.stack(out, axis=1).astype(np.float32)

def augment_biosignal(seq):
    # seq shape: T x C

    # 1. Gaussian noise
    if random.random() < 0.8:
        noise = np.random.normal(0, 0.03, seq.shape).astype(np.float32)
        seq = seq + noise

    # 2. Amplitude scaling
    if random.random() < 0.7:
        scale = np.random.uniform(0.90, 1.10, size=(1, seq.shape[1])).astype(np.float32)
        seq = seq * scale

    # 3. Time shift
    if random.random() < 0.5:
        shift = random.randint(-8, 8)
        seq = np.roll(seq, shift, axis=0)

    # 4. Random masking/dropout of tiny parts
    if random.random() < 0.3:
        t = random.randint(0, max(0, seq.shape[0] - 1))
        width = random.randint(2, 8)
        seq[t:t+width] = 0

    return seq.astype(np.float32)

def get_bio_array(row, train=True):
    start = int(row["bio_start"])
    end = int(row["bio_end"])

    seq = bio_values[start:end]
    seq = (seq - BIO_MEAN) / BIO_STD
    seq = resample_bio_sequence(seq, BIO_SEQ_LEN)

    if train:
        seq = augment_biosignal(seq)

    # return C x T
    return seq.T.astype(np.float32)

# ============================================================
# DATASET
# ============================================================

class AugmentedBioDataset(Dataset):
    def __init__(self, df, train=True, augment_factor=1):
        self.df = df.reset_index(drop=True)
        self.train = train
        self.augment_factor = augment_factor if train else 1

    def __len__(self):
        return len(self.df) * self.augment_factor

    def __getitem__(self, idx):
        real_idx = idx % len(self.df)
        row = self.df.iloc[real_idx]

        x = get_bio_array(row, train=self.train)
        y = int(row["bio_label"])

        return torch.tensor(x, dtype=torch.float32), torch.tensor(y, dtype=torch.long)

# ============================================================
# MODEL
# ============================================================

class Bio1DCNN_Augmented(nn.Module):
    def __init__(self, in_channels=6, num_classes=7):
        super().__init__()

        self.net = nn.Sequential(
            nn.Conv1d(in_channels, 64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(64, 128, kernel_size=5, padding=2),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm1d(128),
            nn.ReLU(),

            nn.Conv1d(128, 64, kernel_size=3, padding=1),
            nn.BatchNorm1d(64),
            nn.ReLU(),

            nn.AdaptiveAvgPool1d(1)
        )

        self.classifier = nn.Sequential(
            nn.Dropout(0.35),
            nn.Linear(64, num_classes)
        )

    def forward(self, x):
        x = self.net(x).squeeze(-1)
        return self.classifier(x)

# ============================================================
# DATALOADERS WITH BALANCED SAMPLING
# ============================================================

train_labels = train_df["bio_label"].astype(int).values
counts = np.bincount(train_labels, minlength=NUM_CLASSES)
sample_weights = np.array([1.0 / counts[y] for y in train_labels], dtype=np.float32)

# because dataset is repeated by augment_factor, repeat weights too
AUGMENT_FACTOR = 4
sample_weights = np.tile(sample_weights, AUGMENT_FACTOR)

sampler = WeightedRandomSampler(
    weights=sample_weights,
    num_samples=len(sample_weights),
    replacement=True
)

train_dataset = AugmentedBioDataset(train_df, train=True, augment_factor=AUGMENT_FACTOR)
test_dataset = AugmentedBioDataset(test_df, train=False)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    sampler=sampler,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

# ============================================================
# TRAINING
# ============================================================

def class_weights(labels, num_classes):
    counts = np.bincount(labels, minlength=num_classes).astype(np.float32)
    counts[counts == 0] = 1.0
    weights = counts.sum() / (num_classes * counts)
    return torch.tensor(weights, dtype=torch.float32).to(DEVICE)

model = Bio1DCNN_Augmented(in_channels=len(BIO_COLUMNS), num_classes=NUM_CLASSES).to(DEVICE)

criterion = nn.CrossEntropyLoss(
    weight=class_weights(train_labels, NUM_CLASSES),
    label_smoothing=0.03
)

optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=5
)

best_acc = 0
best_state = None

for epoch in range(1, EPOCHS + 1):
    model.train()
    losses = []

    for xb, yb in train_loader:
        xb = xb.to(DEVICE)
        yb = yb.to(DEVICE)

        optimizer.zero_grad()
        logits = model(xb)
        loss = criterion(logits, yb)
        loss.backward()
        optimizer.step()

        losses.append(loss.item())

    # eval
    model.eval()
    y_true = []
    y_pred = []

    with torch.no_grad():
        for xb, yb in test_loader:
            xb = xb.to(DEVICE)
            logits = model(xb)
            pred = logits.argmax(dim=1).cpu().numpy()

            y_true.extend(yb.numpy().tolist())
            y_pred.extend(pred.tolist())

    acc = accuracy_score(y_true, y_pred)
    scheduler.step(acc)

    print(f"Epoch {epoch:02d}/{EPOCHS} | Loss {np.mean(losses):.4f} | Val Acc {acc:.4f}")

    if acc > best_acc:
        best_acc = acc
        best_state = copy.deepcopy(model.state_dict())

model.load_state_dict(best_state)

# ============================================================
# FINAL REPORT
# ============================================================

model.eval()
y_true = []
y_pred = []

with torch.no_grad():
    for xb, yb in test_loader:
        xb = xb.to(DEVICE)
        logits = model(xb)
        pred = logits.argmax(dim=1).cpu().numpy()

        y_true.extend(yb.numpy().tolist())
        y_pred.extend(pred.tolist())

target_names = [
    f"{ID_TO_EMOTION[i]}-{EMOTION_NAMES[ID_TO_EMOTION[i]]}"
    for i in range(NUM_CLASSES)
]

print("\nFINAL AUGMENTED BIOSIGNAL REPORT")
print("Best accuracy:", best_acc)
print("Final accuracy:", accuracy_score(y_true, y_pred))

print(classification_report(
    y_true,
    y_pred,
    labels=list(range(NUM_CLASSES)),
    target_names=target_names,
    zero_division=0
))

print("Confusion matrix:")
print(confusion_matrix(y_true, y_pred, labels=list(range(NUM_CLASSES))))

# ============================================================
# SAVE MODEL
# ============================================================

save_path = os.path.join(OUT_DIR, "bio_1dcnn_emotion_augmented.pt")

torch.save({
    "model_state_dict": model.state_dict(),
    "emotion_map": EMOTION_MAP,
    "bio_columns": BIO_COLUMNS,
    "bio_seq_len": BIO_SEQ_LEN,
    "bio_mean_path": "bio_mean_augmented.npy",
    "bio_std_path": "bio_std_augmented.npy",
    "model_type": "augmented_cnn1d_biosignal_emotion"
}, save_path)

print("\nSaved augmented biosignal model:")
print(save_path)
```

```python
# ============================================================
# BIOSIGNAL DIRECT SLIDING-WINDOW TRAINING
# Uses full 4-Hertz.csv directly:
# BVP, EDA, TEMP, X, Y, Z window -> EMOTION
# Saves: bio_1dcnn_sliding_window.pt
# ============================================================

!pip install -q scikit-learn

import os
import json
import copy
import random
import warnings
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F

from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler

warnings.filterwarnings("ignore")

# ============================================================
# CONFIG
# ============================================================

BIO_CSV = "/content/drive/MyDrive/NeuroBioSense_Dataset/NeuroBioSense/Biosignal Files/Pre-Processed/4-Hertz.csv"
OUT_DIR = "/content/drive/MyDrive/NeuroBioSense_Outputs"

BIO_COLUMNS = ["BVP", "EDA", "TEMP", "X", "Y", "Z"]

WINDOW = 64        # 64 rows = 16 seconds at 4Hz
STRIDE = 4         # move by 1 second
BATCH_SIZE = 128
EPOCHS = 25
LR = 1e-3
SEED = 42
NUM_CLASSES = 7

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

EMOTION_MAP = {
    "N": 0,
    "A": 1,
    "D": 2,
    "F": 3,
    "J": 4,
    "SA": 5,
    "SU": 6,
}

ID_TO_EMOTION = {v: k for k, v in EMOTION_MAP.items()}

EMOTION_NAMES = {
    "N": "Neutral",
    "A": "Anger",
    "D": "Disgust",
    "F": "Fear",
    "J": "Joy",
    "SA": "Sadness",
    "SU": "Surprise",
}

# ============================================================
# LOAD BIOSIGNAL CSV
# ============================================================

bio_df = pd.read_csv(BIO_CSV)

print("CSV shape:", bio_df.shape)
print("Columns:", bio_df.columns.tolist())

missing = [c for c in BIO_COLUMNS + ["EMOTION"] if c not in bio_df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

bio_df = bio_df.dropna(subset=BIO_COLUMNS + ["EMOTION"]).reset_index(drop=True)

X_raw = bio_df[BIO_COLUMNS].values.astype(np.float32)
y_raw = bio_df["EMOTION"].map(EMOTION_MAP).values.astype(np.int64)

print("Raw label counts:", Counter(y_raw))

# ============================================================
# CREATE SLIDING WINDOWS
# ============================================================

def make_windows(X, y, window=64, stride=4):
    Xw = []
    yw = []

    for start in range(0, len(X) - window + 1, stride):
        end = start + window
        x_win = X[start:end]
        y_win = y[start:end]

        # dominant emotion in window
        label = int(np.bincount(y_win, minlength=NUM_CLASSES).argmax())

        Xw.append(x_win)
        yw.append(label)

    return np.array(Xw, dtype=np.float32), np.array(yw, dtype=np.int64)

Xw, yw = make_windows(X_raw, y_raw, WINDOW, STRIDE)

print("Windowed X:", Xw.shape)  # N x T x C
print("Windowed y:", yw.shape)
print("Window label counts:", Counter(yw))

# ============================================================
# TRAIN/TEST SPLIT
# Stratified split by emotion
# ============================================================

idx = np.arange(len(yw))

train_idx, test_idx = train_test_split(
    idx,
    test_size=0.2,
    random_state=SEED,
    stratify=yw
)

X_train = Xw[train_idx]
X_test = Xw[test_idx]
y_train = yw[train_idx]
y_test = yw[test_idx]

print("Train:", X_train.shape, Counter(y_train))
print("Test:", X_test.shape, Counter(y_test))

# ============================================================
# NORMALIZE USING TRAIN ONLY
# ============================================================

mean = X_train.reshape(-1, X_train.shape[-1]).mean(axis=0, keepdims=True)
std = X_train.reshape(-1, X_train.shape[-1]).std(axis=0, keepdims=True) + 1e-6

X_train = (X_train - mean) / std
X_test = (X_test - mean) / std

np.save(os.path.join(OUT_DIR, "bio_sliding_mean.npy"), mean)
np.save(os.path.join(OUT_DIR, "bio_sliding_std.npy"), std)

# Convert N x T x C -> N x C x T for Conv1D
X_train = np.transpose(X_train, (0, 2, 1))
X_test = np.transpose(X_test, (0, 2, 1))

# ============================================================
# DATASET
# ============================================================

class BioWindowDataset(Dataset):
    def __init__(self, X, y, train=False):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.long)
        self.train = train

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        x = self.X[idx].clone()
        y = self.y[idx]

        # light train-only augmentation
        if self.train:
            if random.random() < 0.5:
                x = x + torch.randn_like(x) * 0.02
            if random.random() < 0.4:
                scale = torch.empty(x.shape[0], 1).uniform_(0.95, 1.05)
                x = x * scale

        return x, y

# ============================================================
# BALANCED SAMPLER
# ============================================================

counts = np.bincount(y_train, minlength=NUM_CLASSES)
sample_weights = np.array([1.0 / counts[y] for y in y_train], dtype=np.float32)

sampler = WeightedRandomSampler(
    sample_weights,
    num_samples=len(sample_weights),
    replacement=True
)

train_loader = DataLoader(
    BioWindowDataset(X_train, y_train, train=True),
    batch_size=BATCH_SIZE,
    sampler=sampler,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    BioWindowDataset(X_test, y_test, train=False),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

# ============================================================
# MODEL
# ============================================================

class BioSlidingCNN1D(nn.Module):
    def __init__(self, in_channels=6, num_classes=7):
        super().__init__()

        self.net = nn.Sequential(
            nn.Conv1d(in_channels, 64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(64, 128, kernel_size=5, padding=2),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm1d(128),
            nn.ReLU(),

            nn.AdaptiveAvgPool1d(1)
        )

        self.classifier = nn.Sequential(
            nn.Dropout(0.30),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        x = self.net(x).squeeze(-1)
        return self.classifier(x)

model = BioSlidingCNN1D(
    in_channels=len(BIO_COLUMNS),
    num_classes=NUM_CLASSES
).to(DEVICE)

# ============================================================
# TRAIN
# ============================================================

def class_weights(labels, num_classes):
    counts = np.bincount(labels, minlength=num_classes).astype(np.float32)
    counts[counts == 0] = 1.0
    weights = counts.sum() / (num_classes * counts)
    return torch.tensor(weights, dtype=torch.float32).to(DEVICE)

criterion = nn.CrossEntropyLoss(
    weight=class_weights(y_train, NUM_CLASSES),
    label_smoothing=0.02
)

optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-4)

best_acc = 0
best_state = None

for epoch in range(1, EPOCHS + 1):
    model.train()
    losses = []

    for xb, yb in train_loader:
        xb = xb.to(DEVICE)
        yb = yb.to(DEVICE)

        optimizer.zero_grad()
        logits = model(xb)
        loss = criterion(logits, yb)
        loss.backward()
        optimizer.step()

        losses.append(loss.item())

    model.eval()
    y_true = []
    y_pred = []

    with torch.no_grad():
        for xb, yb in test_loader:
            xb = xb.to(DEVICE)
            logits = model(xb)
            pred = logits.argmax(dim=1).cpu().numpy()

            y_true.extend(yb.numpy().tolist())
            y_pred.extend(pred.tolist())

    acc = accuracy_score(y_true, y_pred)

    print(f"Epoch {epoch:02d}/{EPOCHS} | Loss {np.mean(losses):.4f} | Val Acc {acc:.4f}")

    if acc > best_acc:
        best_acc = acc
        best_state = copy.deepcopy(model.state_dict())

model.load_state_dict(best_state)

# ============================================================
# FINAL REPORT
# ============================================================

model.eval()
y_true = []
y_pred = []

with torch.no_grad():
    for xb, yb in test_loader:
        xb = xb.to(DEVICE)
        logits = model(xb)
        pred = logits.argmax(dim=1).cpu().numpy()

        y_true.extend(yb.numpy().tolist())
        y_pred.extend(pred.tolist())

target_names = [
    f"{ID_TO_EMOTION[i]}-{EMOTION_NAMES[ID_TO_EMOTION[i]]}"
    for i in range(NUM_CLASSES)
]

print("\nFINAL SLIDING-WINDOW BIOSIGNAL REPORT")
print("Best accuracy:", best_acc)
print("Final accuracy:", accuracy_score(y_true, y_pred))

print(classification_report(
    y_true,
    y_pred,
    labels=list(range(NUM_CLASSES)),
    target_names=target_names,
    zero_division=0
))

print("Confusion matrix:")
print(confusion_matrix(y_true, y_pred, labels=list(range(NUM_CLASSES))))

# ============================================================
# SAVE
# ============================================================

save_path = os.path.join(OUT_DIR, "bio_1dcnn_sliding_window.pt")

torch.save({
    "model_state_dict": model.state_dict(),
    "emotion_map": EMOTION_MAP,
    "bio_columns": BIO_COLUMNS,
    "window": WINDOW,
    "stride": STRIDE,
    "bio_mean_path": "bio_sliding_mean.npy",
    "bio_std_path": "bio_sliding_std.npy",
    "model_type": "sliding_window_cnn1d_biosignal_emotion"
}, save_path)

print("\nSaved:")
print(save_path)
print(os.path.join(OUT_DIR, "bio_sliding_mean.npy"))
print(os.path.join(OUT_DIR, "bio_sliding_std.npy"))
```

```python
# ============================================================
# FAST 3-CLASS FUSION IMPROVEMENT
# Positive / Negative / Neutral grouping
# Goal: improve fusion accuracy and make dissonance more robust
# ============================================================

!pip install -q scikit-learn xgboost

import os
import json
import copy
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from collections import Counter
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, balanced_accuracy_score, f1_score
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# ============================================================
# PATHS
# ============================================================

OUT_DIR = "/content/drive/MyDrive/NeuroBioSense_Outputs"
LINK_CSV = os.path.join(OUT_DIR, "linking_table_three_model.csv")

# Optional cached predictions will be saved here
SAVE_CSV = os.path.join(OUT_DIR, "fusion_3class_training_table.csv")

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

# ============================================================
# LOAD LINK TABLE
# ============================================================

df = pd.read_csv(LINK_CSV)

print("Loaded:", df.shape)
print(df.head())

# ============================================================
# MAP 7 EMOTIONS TO 3 GROUPS
# 0 = Neutral
# 1 = Positive
# 2 = Negative
# ============================================================

# Original labels:
# N=0, A=1, D=2, F=3, J=4, SA=5, SU=6

def to_group(label):
    label = int(label)

    if label == 0:
        return 0  # Neutral

    if label in [4, 6]:
        return 1  # Positive: Joy, Surprise

    if label in [1, 2, 3, 5]:
        return 2  # Negative: Anger, Disgust, Fear, Sadness

    return 0

GROUP_NAMES = ["Neutral", "Positive", "Negative"]

df["video_group"] = df["video_label"].apply(to_group)
df["bio_group"] = df["bio_label"].apply(to_group)
df["diff_group"] = (df["video_group"] != df["bio_group"]).astype(int)

print("Video group counts:", Counter(df["video_group"]))
print("Bio group counts:", Counter(df["bio_group"]))
print("Fusion group match/mismatch:", Counter(df["diff_group"]))

# ============================================================
# SIMPLE FEATURE ENGINEERING FROM EXISTING TABLE
# Since we are improving fusion fast, use metadata + labels/groups
# This is for analysis-level fusion. For deployment, frontend already computes groups.
# ============================================================

features = pd.DataFrame()

features["video_label"] = df["video_label"].astype(int)
features["bio_label"] = df["bio_label"].astype(int)

features["video_group"] = df["video_group"].astype(int)
features["bio_group"] = df["bio_group"].astype(int)

features["same_exact_emotion"] = (df["video_label"] == df["bio_label"]).astype(int)
features["same_group"] = (df["video_group"] == df["bio_group"]).astype(int)

features["video_positive"] = (df["video_group"] == 1).astype(int)
features["video_negative"] = (df["video_group"] == 2).astype(int)
features["video_neutral"] = (df["video_group"] == 0).astype(int)

features["bio_positive"] = (df["bio_group"] == 1).astype(int)
features["bio_negative"] = (df["bio_group"] == 2).astype(int)
features["bio_neutral"] = (df["bio_group"] == 0).astype(int)

features["duration_sec"] = df["duration_sec"].astype(float)
features["bio_slice_len"] = (df["bio_end"] - df["bio_start"]).astype(float)

X = features.values.astype(np.float32)
y = df["diff_group"].values.astype(int)
groups = df["video_id"].values

# ============================================================
# GROUPED SPLIT
# ============================================================

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=SEED)
train_idx, test_idx = next(gss.split(X, y, groups=groups))

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

print("Train labels:", Counter(y_train))
print("Test labels:", Counter(y_test))

# ============================================================
# BASELINE RULE
# ============================================================

rule_pred = 1 - X_test[:, features.columns.get_loc("same_group")].astype(int)

print("\nRULE-BASED 3-GROUP FUSION")
print("Accuracy:", accuracy_score(y_test, rule_pred))
print("Balanced Accuracy:", balanced_accuracy_score(y_test, rule_pred))
print("Macro F1:", f1_score(y_test, rule_pred, average="macro"))
print(classification_report(y_test, rule_pred, target_names=["Match", "Mismatch"]))
print("Confusion matrix:")
print(confusion_matrix(y_test, rule_pred))

# ============================================================
# RANDOM FOREST FUSION
# ============================================================

rf = RandomForestClassifier(
    n_estimators=300,
    max_depth=6,
    min_samples_leaf=3,
    class_weight="balanced",
    random_state=SEED
)

rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)

print("\nRANDOM FOREST 3-GROUP FUSION")
print("Accuracy:", accuracy_score(y_test, rf_pred))
print("Balanced Accuracy:", balanced_accuracy_score(y_test, rf_pred))
print("Macro F1:", f1_score(y_test, rf_pred, average="macro"))
print(classification_report(y_test, rf_pred, target_names=["Match", "Mismatch"]))
print("Confusion matrix:")
print(confusion_matrix(y_test, rf_pred))

# ============================================================
# XGBOOST FUSION
# ============================================================

neg, pos = np.bincount(y_train, minlength=2)
scale_pos_weight = neg / max(pos, 1)

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.03,
    subsample=0.9,
    colsample_bytree=0.9,
    eval_metric="logloss",
    random_state=SEED,
    scale_pos_weight=scale_pos_weight
)

xgb.fit(X_train, y_train)
xgb_prob = xgb.predict_proba(X_test)[:, 1]

# Tune threshold for best macro F1
best_thr = 0.5
best_macro = -1

for thr in np.linspace(0.1, 0.9, 81):
    pred = (xgb_prob >= thr).astype(int)
    macro = f1_score(y_test, pred, average="macro")
    if macro > best_macro:
        best_macro = macro
        best_thr = thr

xgb_pred = (xgb_prob >= best_thr).astype(int)

print("\nXGBOOST 3-GROUP FUSION")
print("Best threshold:", best_thr)
print("Accuracy:", accuracy_score(y_test, xgb_pred))
print("Balanced Accuracy:", balanced_accuracy_score(y_test, xgb_pred))
print("Macro F1:", f1_score(y_test, xgb_pred, average="macro"))
print(classification_report(y_test, xgb_pred, target_names=["Match", "Mismatch"]))
print("Confusion matrix:")
print(confusion_matrix(y_test, xgb_pred))

# ============================================================
# SAVE BEST RESULT TABLE + MODEL
# ============================================================

result_df = df.copy()
result_df["video_group_name"] = result_df["video_group"].map(lambda x: GROUP_NAMES[int(x)])
result_df["bio_group_name"] = result_df["bio_group"].map(lambda x: GROUP_NAMES[int(x)])
result_df["diff_group_label"] = result_df["diff_group"].map({0: "Match", 1: "Mismatch"})

result_df.to_csv(SAVE_CSV, index=False)
print("\nSaved table:", SAVE_CSV)

import joblib

rf_path = os.path.join(OUT_DIR, "fusion_3class_random_forest.joblib")
xgb_path = os.path.join(OUT_DIR, "fusion_3class_xgboost.joblib")

joblib.dump(rf, rf_path)
joblib.dump(xgb, xgb_path)

print("Saved RF:", rf_path)
print("Saved XGB:", xgb_path)

print("\nFeature columns used:")
print(features.columns.tolist())
```

```python
import os
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from collections import Counter

OUT_DIR = "/content/drive/MyDrive/NeuroBioSense_Outputs"
LINK_CSV = os.path.join(OUT_DIR, "linking_table_three_model.csv")

df = pd.read_csv(LINK_CSV)

df["exact_diff"] = (df["video_label"].astype(int) != df["bio_label"].astype(int)).astype(int)

# Emotion labels:
# 0=N, 1=A, 2=D, 3=F, 4=J, 5=SA, 6=SU

grouping_variants = {
    "A_original": {
        0: 0,  # Neutral
        4: 1, 6: 1,  # Positive: J, SU
        1: 2, 2: 2, 3: 2, 5: 2  # Negative
    },

    "B_surprise_neutral": {
        0: 0, 6: 0,  # Neutral/uncertain: N, SU
        4: 1,        # Positive: J
        1: 2, 2: 2, 3: 2, 5: 2
    },

    "C_surprise_negative": {
        0: 0,
        4: 1,
        1: 2, 2: 2, 3: 2, 5: 2, 6: 2
    },

    "D_sad_neutral": {
        0: 0, 5: 0,
        4: 1, 6: 1,
        1: 2, 2: 2, 3: 2
    },

    "E_fear_neutral": {
        0: 0, 3: 0,
        4: 1, 6: 1,
        1: 2, 2: 2, 5: 2
    },

    "F_four_groups_surprise_own": {
        0: 0,
        4: 1,
        6: 2,
        1: 3, 2: 3, 3: 3, 5: 3
    },
}

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
_, test_idx = next(gss.split(df, groups=df["video_id"]))
test_df = df.iloc[test_idx].copy()

results = []

for name, mapping in grouping_variants.items():
    temp = test_df.copy()

    temp["video_group"] = temp["video_label"].map(mapping)
    temp["bio_group"] = temp["bio_label"].map(mapping)

    temp["pred_diff"] = (temp["video_group"] != temp["bio_group"]).astype(int)

    y_true = temp["exact_diff"].values
    y_pred = temp["pred_diff"].values

    acc = accuracy_score(y_true, y_pred)
    results.append((name, acc))

    print("\n", "=" * 60)
    print(name)
    print("Accuracy:", acc)
    print("Pred counts:", Counter(y_pred))
    print(classification_report(
        y_true,
        y_pred,
        labels=[0, 1],
        target_names=["Match", "Mismatch"],
        zero_division=0
    ))
    print("Confusion matrix:")
    print(confusion_matrix(y_true, y_pred, labels=[0, 1]))

print("\nBEST GROUPING:")
for name, acc in sorted(results, key=lambda x: x[1], reverse=True):
    print(name, acc)
```

```python
# ============================================================
# FINAL RULE-BASED FUSION EVALUATION
# Emotion grouping for dissonance and purchase-intent analysis
# ============================================================

import os
import pandas as pd
from collections import Counter

from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# PATH
# ============================================================

OUT_DIR = "/content/drive/MyDrive/NeuroBioSense_Outputs"
LINK_CSV = os.path.join(OUT_DIR, "linking_table_three_model.csv")

df = pd.read_csv(LINK_CSV)

print("Loaded linking table:", df.shape)

# ============================================================
# ORIGINAL 7-CLASS LABELS
# ============================================================
# Label mapping:
# 0 = Neutral
# 1 = Anger
# 2 = Disgust
# 3 = Fear
# 4 = Joy
# 5 = Sadness
# 6 = Surprise

# Original exact match/mismatch label
# 0 = Match
# 1 = Mismatch
df["exact_diff"] = (
    df["video_label"].astype(int) != df["bio_label"].astype(int)
).astype(int)

print("Original exact match/mismatch counts:")
print(Counter(df["exact_diff"]))

# ============================================================
# FINAL AFFECTIVE GROUPING
# ============================================================
# For fusion and purchase-intent analysis:
#
# Neutral:
#   N
#
# Positive:
#   J
#
# Negative / high-arousal dissonant:
#   A, D, F, SA, SU
#
# Surprise is treated as negative/high-arousal for purchase-intent
# interpretation because it may indicate shock, uncertainty, or
# unexpected reaction rather than clear positive acceptance.
# ============================================================

def to_affective_group(label: int) -> int:
    label = int(label)

    if label == 0:
        return 0  # Neutral

    if label == 4:
        return 1  # Positive: Joy

    return 2      # Negative / high-arousal: A, D, F, SA, SU


GROUP_NAMES = {
    0: "Neutral",
    1: "Positive",
    2: "Negative / High-Arousal"
}

df["video_group"] = df["video_label"].apply(to_affective_group)
df["bio_group"] = df["bio_label"].apply(to_affective_group)

# Rule-based fusion prediction:
# If video group and biosignal group are same -> Match
# If different -> Mismatch
df["fusion_pred"] = (
    df["video_group"] != df["bio_group"]
).astype(int)

print("\nVideo affective group counts:")
print(Counter(df["video_group"]))

print("\nBiosignal affective group counts:")
print(Counter(df["bio_group"]))

print("\nFusion prediction counts:")
print(Counter(df["fusion_pred"]))

# ============================================================
# GROUPED TEST SPLIT
# Ensures same video does not appear in both train/test.
# Only test split is used for reporting final fusion accuracy.
# ============================================================

gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

_, test_idx = next(
    gss.split(df, groups=df["video_id"])
)

test_df = df.iloc[test_idx].reset_index(drop=True)

y_true = test_df["exact_diff"].values
y_pred = test_df["fusion_pred"].values

# ============================================================
# FINAL METRICS
# ============================================================

acc = accuracy_score(y_true, y_pred)
bal_acc = balanced_accuracy_score(y_true, y_pred)
macro_f1 = f1_score(y_true, y_pred, average="macro")
weighted_f1 = f1_score(y_true, y_pred, average="weighted")

print("\n" + "=" * 60)
print("FINAL RULE-BASED FUSION LAYER RESULTS")
print("=" * 60)

print(f"Accuracy: {acc:.4f} ({acc * 100:.2f}%)")
print(f"Balanced Accuracy: {bal_acc:.4f} ({bal_acc * 100:.2f}%)")
print(f"Macro F1-score: {macro_f1:.4f}")
print(f"Weighted F1-score: {weighted_f1:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_true,
        y_pred,
        labels=[0, 1],
        target_names=["Match", "Mismatch"],
        zero_division=0
    )
)

print("Confusion Matrix:")
cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
print(cm)

print("\nInterpretation:")
print(
    "The rule-based fusion layer maps the seven emotion classes into "
    "three affective groups: Neutral, Positive, and Negative/High-Arousal. "
    "Fusion dissonance is detected when the predicted video affective group "
    "differs from the predicted biosignal affective group. When evaluated "
    "against the original seven-class exact match/mismatch labels, this "
    f"fusion layer achieved {acc * 100:.2f}% accuracy."
)

print("\nImportant Note:")
print(
    "This is not a supervised purchase-intent model. Since the dataset does "
    "not contain purchase/no-purchase labels, the purchase intent output is "
    "an estimated rule-based interpretation derived from emotional dissonance."
)

# ============================================================
# SAVE RESULTS
# ============================================================

save_path = os.path.join(OUT_DIR, "final_rule_based_fusion_results.csv")
test_df.to_csv(save_path, index=False)

print("\nSaved test results to:")
print(save_path)
```
